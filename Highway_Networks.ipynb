```python
import torch
import torch.nn as nn
from torchsummary import summary
```

*the* basic formula is:

y=H(x,WH)∗T(x,WT)+x∗(1−T(x,WT))

```python
class Highway_layer(nn.Module):
    def __init__(self,size):
        super().__init__()
        self.H = nn.Linear(size,size)
        self.T = nn.Linear(size,size)
        self.T.bias.data.fill_(-1)
    def forward(self,input):
        H_out = torch.relu(self.H(input))
        T_out = torch.sigmoid(self.T(input))
        C_out = 1 - T_out
        out = H_out * T_out + input * C_out
        return out
```

```python
class Highway_network(nn.Module):
    def __init__(self,input_size,num_layers,output_size):
        super().__init__()
        self.Input_layer = nn.Linear(input_size,input_size)
        self.Highway_layers = nn.ModuleList(Highway_layer(input_size) for _ in range(num_layers))
        self.Output_layer = nn.Linear(input_size,output_size)
    def forward(self,input):
        x = torch.relu(self.Input_layer(input))
        for layer in self.Highway_layers:
            x = layer(x)
        x = self.Output_layer(x)
        return x
```

```python
model = Highway_network(input_size=784,num_layers=5,output_size=10)
summary(model,input_size=(1,784))
```
